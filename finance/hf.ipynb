```python
import duckdb

# Connect to an in-memory DuckDB database
con = duckdb.connect()
```

```python
# 1. Option A: Query directly using the hf:// protocol (Recommended)
# DuckDB version >= 0.10.3 supports direct Hugging Face path querying
hf_url = "hf://datasets/defeatbeta/yahoo-finance-data/data/US/daily_treasury_yield.parquet"

# query the remote file directly, and save results as a dataframe
df = con.sql(f"""
    SELECT 
        report_date,
        bc_1month,
        bc_3month,
        bc_1year,
        bc_2year,
        bc_5year,
        bc_10year,
        bc_30year
    FROM '{hf_url}'
    ORDER BY report_date DESC
    LIMIT 10
""").df()

df.head()
```

```python
# HTTPS direct download URL
http_url = "https://huggingface.co/datasets/defeatbeta/yahoo-finance-data/resolve/main/data/US/daily_treasury_yield.parquet"

df = con.sql(f"""
    SELECT * 
    FROM read_parquet('{http_url}')
    ORDER BY report_date DESC
""").df()

df.head()
```
