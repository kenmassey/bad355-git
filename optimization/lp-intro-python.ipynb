```python
# I wanted to use the pulp package, but can't get it to work???


# https://developers.google.com/optimization/lp/lp_example
from ortools.linear_solver import pywraplp
```

```python

# Use an integer solver engine like 'SCIP' or 'CBC' instead of 'GLOP'
#   GLOP (Google's Linear Optimization Package) doesn't support IntVar
#solver = pywraplp.Solver.CreateSolver('GLOP')
solver = pywraplp.Solver.CreateSolver('SCIP')

# Decision variables (non-negative)
a = solver.NumVar(0.0, solver.infinity(), 'Product_A')
b = solver.NumVar(0.0, solver.infinity(), 'Product_B')

#a = solver.IntVar(0.0, solver.infinity(), 'Product_A')
#b = solver.IntVar(0.0, solver.infinity(), 'Product_B')

# Labor Constraint: 2.5 * A + 1.0 * B <= 40
solver.Add(2.5 * a + 1.0 * b <= 40.0)
# Machine Time Constraint: 1.0 * A + 2.0 * B <= 50
solver.Add(1.0 * a + 2.0 * b <= 50.0)

# Objective Function: Maximize 5 * A + 7 * B
solver.Maximize(5.0 * a + 7.0 * b)

# Solve the system
status = solver.Solve()

if status == pywraplp.Solver.OPTIMAL:
    print(f"  Units of Product A: {a.solution_value():.2f}")
    print(f"  Units of Product B: {b.solution_value():.2f}")
    print(f"  Maximum Profit: ${solver.Objective().Value():.2f}")
else:
    print("The problem does not have an optimal solution.")
```
