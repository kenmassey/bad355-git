# Hot-n-Ready Pizza

You run a pizza shop that makes only one product - large pepperoni pizzas.

Create a function to simulate the arrival of customers, size of orders,
pizza production, and inventory.  Here are some specs for your simulation.

1. Each time step is 15 seconds.
1. Run for a specified number of minutes (e.g. 4 hours would be 240 minutes, and 960 time steps)
1. At each time step, there should be a probability of a customer arrival (you can assume just 1 additional customer possible per time step).
1. Each customer makes an order for a certain mumber of pizzas, which is a right-skewed random variable.  To start, assume a truncated geometric distribution for up to 20 pizzas in an order, so that the mean is 2.0 pizzas per order.  A real pizza place would keep empirical data on the order sizes.
1. It takes 30 seconds to make a pizza, and only one pizza can be in production at a time.  Once a pizza is produced, it is loaded into the oven, and takes 1 more minute to bake.  The oven can be baking two pizzas at a time.  It takes another 15 seconds to box and load/serve a pizza.

    Therefore it takes 1:45 to make a pizza from start to finish.  Running full steam, a pizza can be completed every 30 seconds.
1. Keep track of the hot-n-ready inventory.  Up to a maximum of 16 pizzas can be stored at a time.  After 30 minutes (from the time taken out of the oven), a pizza must be thrown out because it loses freshness.
1. Most orders can be filled immediately.  If not, put those customers in a queue which should indicate the number of pizzas each customer is waiting for.  Customers are served first-come-first-serve.  For example, if a customer orders 8 pizzas, but only 5 are available, then put that customer in the queue waiting for 3 more.  If another customer arrives wanting only 1 pizza, he/she must wait for the earlier customer's order to be filled first.
1. Use the FIFO method for serving pizzas out of inventory, i.e. take the oldest pizzas out of the warmer.
1. Keep track of the percentage of customers that have to wait, and the average wait time.
1. Keep a record of the average age of pizzas that are served (how long since it came out of the oven).
1. Keep track of the percentage of pizzas that must be thrown out.
1. At each time step, workers decide whether to start a new pizza.  Encode a simple rule, e.g. if the number of pizzas in inventory or production is less than TARGET, then start making a pizza.
1. For the simulation, give a final accounting of how many customers were served, pizzas ordered, pizzas wasted, average freshness, and average inventory.  Also give relevant percentages (e.g. immediate order fill %), and rates (pizzas per hour), etc.
1. Keep track of the inventory history, and plot the time series.
1. Keep track of the number of customers waiting in the queue, and plot the time series.               
    
Make sure that a simulation runs with the correct logic, and that you can vary parameters e.g. customer arrival probability or TARGET.

Now, run 1000 Monte-Carlo simulations for various values of customer arrival probability and TARGET.  
For interesting cases, record stats for immediate order filling and waste.
Demonstrate the trade-off of a higher TARGET inventory: better order fill rates, but potentially more waste.  Comment on optimal TARGETs.

Bonus: modify your simulation to allow multiple customers to arrive at each time step, according to a Poisson distribution.
